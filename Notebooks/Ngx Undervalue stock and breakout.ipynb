```python
# Function to remove outliers using IQR
def remove_outliers_iqr(df, column, factor=1.5):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - factor * IQR
    upper_bound = Q3 + factor * IQR
    return df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]

# Prepare the dataframe
stock_df_cleaned = stock_df2.copy()
stock_df_cleaned = stock_df_cleaned.replace([np.inf, -np.inf], np.nan).dropna(
    subset=["price_earnings_ttm", 
            "earnings_per_share_diluted_yoy_growth_ttm", 
            "market_cap_basic",
            'Candle.3WhiteSoldiers|1M',
            'Candle.Kicking.Bullish|1M',
            'Candle.Engulfing.Bullish|1M',
            'Candle.MorningStar|1M',
            'Candle.Marubozu.White|1M'
           ]
)
stock_df_cleaned = stock_df_cleaned[stock_df_cleaned["market_cap_basic"] > 0]

# Store all undervalued picks
undervalued_all = []

# Iterate through all sectors and industries
for sector in stock_df_cleaned['sector'].dropna().unique():
    sector_df = stock_df_cleaned[stock_df_cleaned['sector'] == sector]
    
    for industry in sector_df['industry'].dropna().unique():
        df = sector_df[sector_df['industry'] == industry].copy()
        
        # Remove outliers
        df = remove_outliers_iqr(df, "price_earnings_ttm", factor=1.5)
        df = remove_outliers_iqr(df, "earnings_per_share_diluted_yoy_growth_ttm", factor=1.5)
        
        if df.empty:
            continue
        
        # Calculate industry average P/E
        avg_pe = df["price_earnings_ttm"].mean()
        
        # Filter undervalued stocks
        undervalued = df[
            (df["price_earnings_ttm"] < avg_pe) &
            (df["earnings_per_share_diluted_yoy_growth_ttm"] > 0)
        ].copy()
        
        if not undervalued.empty:
            undervalued["sector"] = sector
            undervalued["industry"] = industry
            undervalued_all.append(undervalued)

# Combine and sort results
if undervalued_all:
    final_df = pd.concat(undervalued_all)
    final_df = final_df.sort_values(by="market_cap_basic", ascending=False)
    
    display_cols = ["name", "sector", "industry", 
                    "price_earnings_ttm", 
                    "earnings_per_share_diluted_yoy_growth_ttm", 
                    "market_cap_basic",
                    'price_target_1y', 'price_target_1y_delta', 'price_target_average', 'price_target_high', 'price_target_median',
                    'Candle.3WhiteSoldiers|1M', 'Candle.Kicking.Bullish|1M', 
                    'Candle.Engulfing.Bullish|1M', 'Candle.MorningStar|1M', 'Candle.Marubozu.White|1M'
]
    
    picked_tickers = final_df["name"].unique().tolist()
    
    print("✅ Final List of Undervalued Stocks:")
    display(final_df[display_cols].reset_index(drop=True))
    
    print("\n🎯 Tickers Picked:")
    print(picked_tickers)
else:
    print("⚠️ No undervalued stocks found across any sector/industry.")
```
