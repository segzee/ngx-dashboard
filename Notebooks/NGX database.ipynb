```python
# Build the query
query=(Query()
 .select(
     'name',
     'close',
     'basic_eps_net_income',
     'ebitda_fy',
     'Perf.5Y',
     'last_annual_revenue',
     'price_book_ratio',
     'enterprise_value_current',
     'average_volume_30d_calc',
     'enterprise_value_ebitda_ttm',
     'dividends_yield_current'
 )
 .limit(200)
 .set_markets('nigeria')
 .set_property('symbols', {'query': {'types': ['stock', 'dr', 'fund']}})
 )
```
